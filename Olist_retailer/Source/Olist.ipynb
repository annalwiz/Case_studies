### **Context: Olist**
* Olist is established in and operates exclusively within Brazil.
* It is a 100% digital company (business operations are managed entirely online).
* Provides e-commerce services for sellers.
* Connects (small) merchants to Brazil's largest marketplaces (Amazon, Bahia, Walmart, etc.).
* Offers logistics and inventory management services to sellers.
* Does not sell directly to consumers.  
### **Your Mission**  
* **Objective:** Help business leaders in their decision-making process. The Olist CEO has tasked you with making sense of the data in order to formulate recommendations regarding the following question:
    * How can we increase Olist customer satisfaction?
* **The Data**
    * Information on approximately 100,000 orders placed between 2016 and 2018.
    * Real commercial data.
    * Identifiable information has been anonymized.
    * Full documentation is available in the provided data dictionary.
**The Workflow**
*  **The Seller:**
    * Joins Olist and makes their product catalogs available online to Olist.
    * Olist displays these catalogs on existing marketplaces (Amazon, Bahia, Walmart, etc.).
    * The seller receives a notification as soon as a product is sold.
    * The seller hands over the ordered items to third-party logistics carriers.
    * Warning: Multiple sellers may be involved in a single customer order!
* **The Customer:**
    * Browses products on marketplaces (Amazon, Bahia, Walmart, etc.).
    * Purchases products listed via the Olist Store.
    * Receives an estimated delivery date, accounting for the estimated time of arrival.
    * Receives the order(s) and leaves a review.

**Note:** Between 2016 and mid-2018, a review could be left as soon as the order was shipped. This means a customer could potentially leave a review for a product they had not yet received!

## **3. Hypothesis**

#### **3.1 The Main Hypothesis: "The Expectation Gap"**
* **Hypothesis:** Customer satisfaction (Review Score) depends less on the absolute speed of delivery, and more on the accuracy of the delivery promise  
 **`(Estimated Date vs. Actual Date)`**. A customer who waits 20 days but was told to expect 25 will be happier  
 than a customer who waits 5 days but was promised 3.
* If it is the case, then Olist doesn't necessarily need faster trucks (expensive); they need better prediction algorithms (data science).
* This hypothesis leverages specific data: order_estimated_delivery_date and order_delivered_customer_date. 
* It addresses the "Review Anomaly": You noted that reviews can be left before delivery. This suggests that anxiety about a missed deadline drives negative behavior.

```python
# Let's check the null values in orders table : 
# we create a mask to filter rows 
order_df.isna().sum()
mask = order_df['order_delivered_customer_date'].isna() 
order_df[mask].groupby('order_status').size().reset_index(name='counts') # There are 8 orders delivered yet with no delivery date ???? 
# Let's analyse delivered orders only : 
delivered_orders = order_df[order_df['order_status'] == 'delivered']
delivered_orders= delivered_orders.dropna(subset=['order_delivered_customer_date']) # we drop the 8 orders with no delivery date
delivered_orders.info()
```

```python
# Let's augment our table with new useful columns : 
# delivery_time (in days): How long it actually took to deliver the order
# delays (in days): The expectation Gap between the estimated delivery date and the actual delivery date
delivered_orders['delivery_time'] = (delivered_orders['order_delivered_customer_date'] - delivered_orders['order_purchase_timestamp']).dt.days
delivered_orders['delays'] = (delivered_orders['order_delivered_customer_date'] - delivered_orders['order_estimated_delivery_date']).dt.days
```

```python
# Join delivered_orders with reviews_df 
order_review = pd.merge(delivered_orders, reviews_df, on='order_id', how='left')
columns = ['order_id', 'customer_id', 'order_status', 
          'order_delivered_carrier_date', 'order_delivered_customer_date',
          'delivery_time', 'delays', 'review_id', 'review_score', 'review_creation_date']
order_review = order_review[columns] # We keep only useful columns 
order_review

# We add a sentiment analysis column based on review_score
def sentiment_analysis(score):
    if score in [4, 5]:
        return 'positive'
    elif score == 3:
        return 'neutral'
    else:
        return 'negative'
order_review['sentiment'] = order_review['review_score'].apply(sentiment_analysis)
order_review
```

```python
# Group by the new sentiment column
sentiment_analysis = order_review.groupby('sentiment').agg({
    'delays': 'mean',          # The "Promise Gap"
    'delivery_time': 'mean',   # The "Physical Time"
    'review_id': 'count'
}).round(2)

print("--- Impact of Delays on Customer Sentiment ---")
print(sentiment_analysis)
```

* **Everyone is "Early" (On Average):** Even the Negative group received their packages 5.2 days early (Mean Delay = -5.2).
* **Implication:** Olist's estimated dates are extremely conservative (safe). They promise a long wait, so they almost always "beat" the deadline.
Speed Still Matters:
* Happy Customers: Waited ~10.6 days.
* Unhappy Customers: Waited ~19.7 days.
* **Insight:** There is a "Patience Threshold."   
Even if you tell a customer "It will take 25 days" and you deliver it in 20 (beating the estimate),   
they are still unhappy because 20 days is simply too long.

#### **Sellers Analysis**

```python
# 1. Merge the Sellers table to our Master DF
# We use the 'items' table to link orders to sellers
seller_analysis = pd.merge(
    left=order_review,  # Your current table with sentiment
    right=items_df[['order_id', 'seller_id']], # Link Order -> Seller
    on='order_id',
    how='inner'
)

# 2. Aggregation: Find the Sellers with the most unhappy customers
bad_sellers = seller_analysis.groupby('seller_id').agg({
    'review_id': 'count',
    'sentiment': lambda x: (x == 'negative').sum(),
    'delivery_time': 'mean',  # Let's see if they are slow
    'delays': 'mean'          # Let's see if they miss promises
}).rename(columns={'review_id': 'total_orders', 'sentiment': 'bad_reviews'})

# 3. Calculate the "Defect Rate" (%)
bad_sellers['bad_review_rate'] = (bad_sellers['bad_reviews'] / bad_sellers['total_orders']) * 100

# 4. Filter for significance (Sellers with at least 50 orders)
significant_sellers = bad_sellers[bad_sellers['total_orders'] > 50].sort_values('bad_review_rate', ascending=False)

print("--- Top 10 Sellers with Highest Dissatisfaction Rates ---")
significant_sellers.head(10)[['total_orders', 'bad_reviews', 'bad_review_rate', 'delivery_time']]
```

#### **Interpretation**
* The top sellers with bad review rates > 40% and delivery times > 20 days, Olist must kick them off the platform.
* If the delivery times are normal (~10 days) but reviews are still bad, then the problem is Product Quality.

#### **Product Analysis**

```python
# 1. We start by merging Reviews + Items (to link Score -> Product ID)
# We use inner join because we need both the score and the product info
reviews_items = pd.merge(
    left=order_review, # Your current master table (reviews + orders)
    right=items_df, 
    on='order_id', 
    how='inner'
)

# 2. Now merge with the Products table
product_analysis = pd.merge(
    left=reviews_items,
    right=products_df,
    on='product_id',
    how='inner'
)

# 3. Clean up the category names (replace underscores with spaces for readability)
# We assume you might want to translate them later, but for now let's clean the format
product_analysis['product_category_name'] = product_analysis['product_category_name'].str.replace('_', ' ')

print(f"Shape of Product Analysis DF: {product_analysis.shape}")
product_analysis.head(3)
```

#### **Product Category Analysis**

```python
# Group by Category
category_performance = product_analysis.groupby('product_category_name').agg({
    'review_score': 'mean',
    'review_id': 'count',
    'delays': 'mean' # Let's see if bad categories are also just slow!
}).rename(columns={'review_id': 'total_sales'})

# Filter for relevant categories (at least 50 sales) to avoid noise
significant_categories = category_performance[category_performance['total_sales'] > 50].sort_values('review_score')

print("--- The 10 'Most Hated' Product Categories ---")
print(significant_categories.head(10))
```

#### **Interpretation**
* The lowest-rated category was **Office Furniture (Score: 3.5)**, despite arriving ~12 days early.
* **Insight:** The dissatisfaction stems from the Physical vs. Digital disconnect. 
    * Heavy, complex items often arrive with scratches or missing instructions, leading to a gap between the "pristine" digital image and the "complex" physical reality.
